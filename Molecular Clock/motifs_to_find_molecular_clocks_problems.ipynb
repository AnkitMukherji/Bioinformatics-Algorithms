**Brute force algorithm for motif finding**

```python
import time

with open('motif_enumeration.txt', 'r') as f:
    data = f.readlines()
    k, d = data[0].split(' ')
    k = int(k.strip())
    d = int(d.strip())
    Dna = data[1].strip()
    start_time = time.time()
    result = MotifEnumeration(Dna, k, d)
    end_time = time.time()
    print(f"Time taken: {round((end_time - start_time), 2)} seconds")
    with open('motif_enumeration_output.txt', 'w') as out:
        out.write('\n'.join(result))
```

**Profile-most Probable k-mer Problem**

```python
with open('profile_most_probable_kmer.txt', 'r') as f:
    lines = [line.strip() for line in f]
    
    dna = lines[0]
    k = int(lines[1])
    
    profile = []
    for line in lines[2:]:
        profile.append(list(map(float, line.split())))

    print(f"The {k}-mer with the highest probability of {profile_most_probable_kmer(dna, k, profile)[0]} is {profile_most_probable_kmer(dna, k, profile)[1]}")
```
